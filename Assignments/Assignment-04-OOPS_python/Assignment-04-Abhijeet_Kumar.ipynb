```python
class Patron:
    def __init__(self, name, patron_id):
        self.name = name
        self.patron_id = patron_id
        self.books = [] 

    def borrow_book(self, book):
        
        self.books.append(book)
        book.check_out()
        
```

```python
class Library:
    def __init__(self, name, Book, Patron):
        self.name = name
        self.Book = {}
        self.Patron = {}


    def add_book(self, book):
        
        if(book.ISBN in self.Book):
            print(f"{book.ISBN} is already added")
        else:
            self.Book[book.ISBN] = book       
            

    def add_patron(self,patron):
        if(patron.patron_id in self.Patron):
             print("Patron already added")
        else:
            self.Patron[patron.patron_id] = patron   


    def lend_book(self, book_isbn, patron_id):

        book = self.Book.get(book_isbn)
        patron = self.Patron.get(patron_id)
        print(patron)
        if(book.is_available()):
            patron.borrow_book(book)
        else:
            print(f"Book {book.ISBN} is already lent")    

    def return_book(self, book_isbn):
        self.Book[book_isbn].check_in()
```

## Problem-2

Create an base class `Shape` with an method `area()` and another method `perimeter()`. Then, create classes `Rectangle` and `Circle` that inherit from `Shape` and implement the `area()` method. The `perimeter()` method in `Shape` should raise a `NotImplementedError`. Implement the `perimeter()` method in `Rectangle` and `Circle`.

Test your implementation.

```python
class Shape:
    def __init__(self):
        pass

    def area(self):
        raise Exception("Not Implemented error!")

    def perimeter(self):
        raise Exception("Not Implemented error!")
```

```python
class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth
        super().__init__()

    def area(self):
        return (self.length)*(self.breadth)    

    
```

```python
rectangle = Rectangle(10,5)
print(rectangle.area())
print(rectangle.perimeter())
```

```python
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
        super().__init__()

    def area(self):
        return ((self.radius)*(self.radius)*3.14)    

    
```

```python
circle = Circle(5.0)
print(circle.area())
print(circle.perimeter())
```

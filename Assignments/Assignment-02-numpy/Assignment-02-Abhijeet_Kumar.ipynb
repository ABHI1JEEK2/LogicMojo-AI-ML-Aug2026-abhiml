Create an 8 x 8 array with a checkerboard pattern of zeros and ones.

```python
a = np.zeros((8,8))
checkered = True
for i in range (len(a)):
    checkered = not checkered
    for j in range (len(a[i])):
        if(checkered):
            a[i,j] = 1
            checkered = False
        else :
            checkered = True

print(a)
```

Create a function which creates an n*n array with (i,j) -entry equal to i+j

```python
def summatrix(n):
    a = np.zeros((n,n), dtype=int)
    for i in range (n):
        for j in range (n):
           
            a[i,j] = i+j
    
    return a
result = summatrix(3)
result
```

Evaluate cos and sin on the interval [0,1] with 1000 points and then stack the results into a tall array with rows being the (cos(x),sinx(x)) entries

```python
x= np.linspace(0,1,1000)
x
a = np.sin(x)
print(a.shape)
b = np.cos(x)
print(b.shape)
c = np.vstack((a,b))
print(np.transpose(c))
```

11. Add a column that calculates the percent change between the 2010 Census and the 2017 Population Estimate

```python
population_pct = ((pop['2017PopEstimate']-pop['2010Census'])*100)/pop['2010Census']
pop['population_pct'] = population_pct
pop.head()
```
